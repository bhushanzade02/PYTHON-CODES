```python
import pandas as pd
import numpy as np
products = pd.read_csv("products.csv")
transactions = pd.read_csv("transactions.csv")
```

(a) DATA CLEANING

1️ Handle missing values in price column

```python
products['price'].isna().sum()
```

##
Why mean?

Price is numerical

Mean keeps overall price distribution balanced

```python
mean_price = products['price'].mean()
products['price'].fillna(mean_price, inplace=True)
```

2 Remove duplicate transaction records

transactions = transactions.drop_duplicates()


```python
transactions = transactions.drop_duplicates()
```

(b) DATA MERGING & ANALYSIS

3 Merge datasets on product_id

```python
merged_df = pd.merge(transactions, products, on='product_id', how='inner')
```

4 Total quantity sold per car segment

```python
segment_quantity = merged_df.groupby('segment')['quantity'].sum()
print(segment_quantity)
```

(c) INTERPRETATION & INSIGHTS

5️ Segment with highest total quantity sold

```python
highest_segment = segment_quantity.idxmax()
print(highest_segment)
```

6️ Least sold brand

```python
brand_sales = merged_df.groupby('brand')['quantity'].sum()
least_sold_brand = brand_sales.idxmin()
print(least_sold_brand)
```

7 Top 3 months with highest sales

```python
# Convert date column:
merged_df['transaction_date'] = pd.to_datetime(merged_df['transaction_date'])
#Extract month:
merged_df['month'] = merged_df['transaction_date'].dt.month

```

```python
# Group by month:
top_months = merged_df.groupby('month')['quantity'].sum().sort_values(ascending=False).head(3)
print(top_months)
```

CAR PRICE ANALYSIS

```python
total_price = products['price'].sum()
missing_price = products['price'].isna().sum()
missing_percentage = (missing_price / len(products)) * 100
min_price = products['price'].min()
max_price = products['price'].max()
mean_price = products['price'].mean()
median_price = products['price'].median()
print("Total Price:", total_price)
print("Missing Prices:", missing_price)
print("Missing %:", missing_percentage)
print("Min Price:", min_price)
print("Max Price:", max_price)
print("Mean Price:", mean_price)
print("Median Price:", median_price)
```
